1. Kiểm tra số dòng và cột của các bảng

```python
import pandas as pd

DATA_DIR = "../../data/raw/fma_metadata/"

tracks = pd.read_csv(DATA_DIR + "tracks.csv", header=[0, 1], index_col=0, low_memory=False)
features = pd.read_csv(DATA_DIR + "features.csv", header=[0, 1, 2], index_col=0)
echonest = pd.read_csv(DATA_DIR + "echonest.csv", header=[0, 1, 2], index_col=0, low_memory=False)
genres = pd.read_csv(DATA_DIR + "genres.csv")


datasets = {
    "Tracks": tracks,
    "Features": features,
    "Echonest": echonest,
    "Genres": genres
}

data = []

for name, df in datasets.items():
    rows = df.shape[0]
    cols = df.shape[1]

    data.append([name, rows, cols])

summary = pd.DataFrame(
    data,
    columns=["Bảng", "Số bản ghi", "Số thuộc tính"]
)

display(summary)
```

3. Check thuộc tính tiêu biểu của features.csv

```python

display(features[[
    "mfcc",
    "chroma_cens",
    "spectral_centroid",
    "spectral_contrast",
    "tonnetz",
    "zcr"
]].head())


selected_features = []

for group in [
    "mfcc",
    "chroma_cens",
    "spectral_centroid",
    "spectral_contrast",
    "tonnetz",
    "zcr"
]:
    # Lấy một cột đại diện trong mỗi nhóm
    col = features[group].columns[0]
    selected_features.append((group, *col))

display(features[selected_features].head())
```
